```python
import os
import glob
import rasterio
import numpy as np
from PIL import Image

def convert_geotiff_to_png_or_jpg(
    input_dir, 
    output_dir="converted_images", 
    out_ext="png",   # or "jpg"
    out_dtype=np.uint8
):
    os.makedirs(output_dir, exist_ok=True)
    tifs = glob.glob(os.path.join(input_dir, "*.tif")) + glob.glob(os.path.join(input_dir, "*.tiff"))

    for tif_path in tifs:
        with rasterio.open(tif_path) as src:
            arr = src.read([1, 2, 3])
            img = np.moveaxis(arr, 0, -1)  # (rows, cols, 3)
            if img.dtype != out_dtype:
                img = (img / img.max() * 255).astype(out_dtype)

        out_name = os.path.splitext(os.path.basename(tif_path))[0] + f".{out_ext}"
        out_path = os.path.join(output_dir, out_name)
        Image.fromarray(img).save(out_path)
        print(f"Saved: {out_path}")

# Example usage:
convert_geotiff_to_png_or_jpg(
    input_dir="test_sample/",         # Input folder of .tif/.tiff images
    output_dir="converted_images",    # Where to put PNGs/JPGs
    out_ext="jpg"                    # Change to "jpg" for JPEG
)
```
