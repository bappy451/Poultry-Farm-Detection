```python
import cv2
import torch
import matplotlib.pyplot as plt
from ultralytics import YOLO
from pathlib import Path
from IPython.display import display, clear_output
from PIL import Image
import numpy as np

# Load the trained model
# model = YOLO("C:/Users/mza0288/Documents/CODE/runs/segment/train27/weights/best.pt")

# Get validation dataset information
# val_results = model.val()
dataset = "C:/data/data_set/test"  # Get the validation dataset path

data_path = Path(dataset)
images_path = data_path / "images"  # Adjust this if needed
labels_path = data_path / "labels"  # Adjust this if needed

# Get list of images
image_files = list(images_path.glob("*.jpg"))  # Adjust extension if needed

# Function to combine ground truth and prediction into a single image
def save_combined_image(img_path, model, save_path):
    img = cv2.imread(str(img_path))
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    
    # Predict
    results = model(img_path)[0]
    pred_img = results.plot()  # Get prediction image as a numpy array
    
    # Load ground truth label
    label_img = img.copy()
    label_path = labels_path / (img_path.stem + ".txt")
    if label_path.exists():
        with open(label_path, "r") as f:
            for line in f:
                values = list(map(float, line.strip().split()))
                class_id = int(values[0])
                mask_points = values[1:]  # Instance segmentation mask points
                mask_points = [(int(mask_points[i] * img.shape[1]), int(mask_points[i+1] * img.shape[0])) for i in range(0, len(mask_points), 2)]
                cv2.polylines(label_img, [np.array(mask_points, np.int32)], isClosed=True, color=(0, 255, 0), thickness=2)
                cv2.putText(label_img, str(class_id), mask_points[0], cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)
    
    # Combine images side by side
    combined_img = np.hstack((label_img, pred_img))
    
    # Convert to PIL and save
    combined_pil = Image.fromarray(combined_img)
    combined_pil.save(save_path / f"{img_path.stem}_comparison.jpg")

# Create save directory
save_dir = Path("C:/Users/mza0288/Documents/CODE/validation_results_100")
save_dir.mkdir(parents=True, exist_ok=True)

# Iterate through images and save results
for img_file in image_files:
    save_combined_image(img_file, model, save_dir)
```
