```python
# Cell 1: Import necessary libraries
import torch
import numpy as np
import cv2
import matplotlib.pyplot as plt
from segment_anything import sam_model_registry, SamPredictor

# Display inline
%matplotlib inline
```

```python
# Cell 2: Load image
image_path = "C:/Users/mza0288/Documents/Code/annotated_output/m_3108501_se_16_030_20211103_annotated.jpg"  # Change this to your image file
image = cv2.imread(image_path)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
# Cell 3: Load the SAM model
sam_checkpoint = "sam_vit_h_4b8939.pth"  # Make sure you download this checkpoint
model_type = "vit_h"

device = "cuda" if torch.cuda.is_available() else "cpu"
sam = sam_model_registry[model_type](checkpoint=sam_checkpoint)
sam.to(device)
predictor = SamPredictor(sam)



# Cell 4: Set image for prediction
predictor.set_image(image)

# Manually select a point (for analysis)
input_point = np.array([[image.shape[1] // 2, image.shape[0] // 2]])  # center
input_label = np.array([1])  # foreground

masks, scores, logits = predictor.predict(
    point_coords=input_point,
    point_labels=input_label,
    multimask_output=True,
)
```

```python
# Cell 5: Show the masks
plt.figure(figsize=(18, 6))
for i, mask in enumerate(masks):
    plt.subplot(1, 3, i+1)
    plt.imshow(image)
    plt.imshow(mask, alpha=0.6)
    plt.scatter(input_point[:, 0], input_point[:, 1], c='red', s=40)
    plt.title(f"Mask {i+1} - Score: {scores[i]:.3f}")
    plt.axis("off")
plt.show()
```
