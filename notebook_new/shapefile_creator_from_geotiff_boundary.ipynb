```python
import os
import rasterio
from shapely.geometry import box, mapping
import fiona

def create_shapefiles_from_geotiffs(geotiff_folder, output_folder):
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    for filename in os.listdir(geotiff_folder):
        if filename.lower().endswith((".tif", ".tiff")):
            filepath = os.path.join(geotiff_folder, filename)
            with rasterio.open(filepath) as src:
                bounds = src.bounds
                crs = src.crs.to_dict()
                schema = {
                    'geometry': 'Polygon',
                    'properties': {'filename': 'str'}
                }

                geom = box(*bounds)
                output_shapefile = os.path.join(output_folder, os.path.splitext(filename)[0] + ".shp")

                with fiona.open(
                    output_shapefile, 'w',
                    driver='ESRI Shapefile',
                    crs=crs,
                    schema=schema
                ) as layer:
                    layer.write({
                        'geometry': mapping(geom),
                        'properties': {'filename': filename}
                    })

    return f"Shapefiles created in: {output_folder}"

# Example usage:
# create_shapefiles_from_geotiffs("/path/to/geotiff/folder", "/path/to/output/shapefiles")
```
