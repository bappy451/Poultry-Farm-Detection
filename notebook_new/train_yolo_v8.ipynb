```python
from ultralytics import YOLO
import torch
import os
import pandas as pd
import wandb
import os

models = ["yolov8n.pt", "yolov8s.pt", "yolov8m.pt", "yolov8l.pt"]
data_path = "C:/data/data_rice/humay-ai-rice-diseases/data.yaml"
project="runs/rice_disease",
run_name="none"

device = "mps" if torch.backends.mps.is_available() else ("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

epochs = 100

for model_name in models:
    print(f"\nTraining model: {model_name}")

    # Clear CUDA cache before starting new training (if using CUDA)
    if device == "cuda":
        torch.cuda.empty_cache()

    model = YOLO(model_name).to(device)
    batch_size = 8
    imgsz = 640

    run_name = f"{model_name.split('.')[0]}-bs{batch_size}-ep{epochs}"
    
    print(f"\nTraining RUN: {run_name}")

    results = model.train(
        data=data_path,
        epochs=epochs,
        batch=batch_size,
        imgsz=imgsz,
        device=device,
        seed=7,
        optimizer="SGD",
        show_boxes=False,
        show_labels=False,
        project="runs/rice_disease",
        name=run_name,
        save=True,
        verbose=True,
    )

    
    # Initialize wandb before training or metrics logging
    wandb.init(project="rice_disease", name=run_name)
    
    # Path to your saved metrics CSV
    metrics_csv_path = f"runs/rice_disease/{run_name}/results.csv"
    
    # Load the CSV into DataFrame
    df = pd.read_csv(metrics_csv_path)
    
    # Iterate over each epoch's metrics row
    for _, row in df.iterrows():
        # Convert epoch to int for clarity
        epoch_num = int(row['epoch'])
    
        # Prepare dictionary of metrics, exclude epoch key
        metrics = {k: float(v) for k, v in row.items() if k != 'epoch'}
    
        # Add epoch number to the dictionary for logging
        metrics['epoch'] = epoch_num
    
        # Log metrics for this epoch to wandb
        wandb.log(metrics)
    
    wandb.finish()
```
