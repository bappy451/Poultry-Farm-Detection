```python
import torch
from torch.utils.data import Dataset, DataLoader
from torchvision.datasets import CocoDetection
from transformers import DetrImageProcessor, DetrForSegmentation
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
from pycocotools.coco import COCO
import torchvision.transforms as T
```

```python
class CocoSegmentationDataset(Dataset):
    def __init__(self, img_folder, ann_file, processor):
        self.img_folder = img_folder
        self.coco = COCO(ann_file)
        self.img_ids = self.coco.getImgIds()
        self.processor = processor

    def __len__(self):
        return len(self.img_ids)

    def __getitem__(self, idx):
        img_id = self.img_ids[idx]
        ann_ids = self.coco.getAnnIds(imgIds=img_id)
        annotations = self.coco.loadAnns(ann_ids)
        image_info = self.coco.loadImgs(img_id)[0]

        image = Image.open(f"{self.img_folder}/{image_info['file_name']}").convert("RGB")

        formatted_annotations = [
            {
                "bbox": ann["bbox"],
                "category_id": ann["category_id"],
                "area": ann.get("area", ann["bbox"][2] * ann["bbox"][3]),
                "iscrowd": ann.get("iscrowd", 0),
                "segmentation": ann["segmentation"]
            } for ann in annotations
        ]

        encoding = self.processor(images=image, annotations={"image_id": img_id, "annotations": formatted_annotations}, return_tensors="pt")
        encoding = {k: v.squeeze(0) for k, v in encoding.items() if isinstance(v, torch.Tensor)}
        encoding["labels"] = [{"image_id": img_id, "annotations": formatted_annotations}]

        return encoding

# Initialize processor and model
processor = DetrImageProcessor.from_pretrained("facebook/detr-resnet-50-panoptic")
model = DetrForSegmentation.from_pretrained("facebook/detr-resnet-50-panoptic")

# Dataset and loader
train_dataset = CocoSegmentationDataset('C:/Users/mza0288/Datset/train2017', 'C:/Users/mza0288/Datset/annotations/instances_train2017.json', processor)
train_loader = DataLoader(train_dataset, batch_size=2, shuffle=True)
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model.to(device)

optimizer = torch.optim.AdamW(model.parameters(), lr=1e-4)

# Training loop
model.train()
epochs = 10

for epoch in range(epochs):
    total_loss = 0
    for batch in train_loader:
        optimizer.zero_grad()

        pixel_values = batch['pixel_values'].to(device)
        labels = batch['labels']

        outputs = model(pixel_values=pixel_values, labels=labels)
        loss = outputs.loss
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    print(f"Epoch [{epoch+1}/{epochs}] - Loss: {total_loss / len(train_loader):.4f}")
```

```python
# Save model
torch.save(model.state_dict(), "detr_segmentation.pth")
```

```python
def predict(image_path):
    model.eval()
    image = Image.open(image_path).convert("RGB")
    inputs = processor(images=image, return_tensors="pt").to(device)

    with torch.no_grad():
        outputs = model(**inputs)

    result = processor.post_process_panoptic_segmentation(outputs, target_sizes=[image.size[::-1]])[0]

    plt.figure(figsize=(10, 10))
    plt.imshow(image)
    plt.imshow(result["segmentation"])
    plt.axis("off")
    plt.show()

# Example usage
predict('C:/Users/mza0288/Datset/val2017/tile_0_0_jpg.rf.2bc2a6231fb40041d7845e1918741a3d.jpg')
```
