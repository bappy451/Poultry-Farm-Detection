```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Data preparation
data = {
    'Dataset': ['Training', 'Validation', 'Testing', 'Training', 'Training', 'Validation', 'Validation', 'Testing', 'Testing'],
    'Category': ['Series1', 'Series1', 'Series1', 'Broiler House', 'Broiler Breeder', 'Broiler House', 'Broiler Breeder', 'Broiler House', 'Broiler Breeder'],
    'Count': [1446, 366, 702, 5109, 527, 1366, 104, 2950, 193]
}

df = pd.DataFrame(data)

# Visualization: Unique representation (Sunburst-like pie chart using nested donut chart)

fig, ax = plt.subplots(figsize=(8, 8))
size = 0.3

# Outer Ring (Dataset and Category)
outer_vals = df['Count']
outer_labels = df['Dataset'] + ' - ' + df['Category']
outer_colors = sns.color_palette("Set3", len(outer_labels))

ax.pie(outer_vals, radius=1, labels=outer_labels, colors=outer_colors,
       wedgeprops=dict(width=size, edgecolor='w'), autopct='%1.1f%%', pctdistance=0.85)

# Inner Ring (Dataset only)
inner_vals = df.groupby('Dataset')['Count'].sum()
inner_labels = inner_vals.index
inner_colors = sns.color_palette("Set2", len(inner_labels))

ax.pie(inner_vals, radius=1-size, labels=inner_labels, labeldistance=0.7, colors=inner_colors,
       wedgeprops=dict(width=size, edgecolor='w'), autopct='%1.1f%%', pctdistance=0.6)

ax.set(aspect="equal", title='Unique Nested Pie Representation of Dataset Distribution')
plt.tight_layout()
plt.show()
```
