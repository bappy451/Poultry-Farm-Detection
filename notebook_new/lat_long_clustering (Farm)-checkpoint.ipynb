```python
import pandas as pd
import numpy as np
import geopandas as gpd
from shapely.geometry import Point
import matplotlib.pyplot as plt
import networkx as nx
from sklearn.neighbors import KDTree

# Step 1: Load and Prepare Data
df = pd.read_csv("csv/combined_file.csv").dropna(subset=['global_x', 'global_y'])
coords = df[['global_x', 'global_y']].values

# Step 2: Build Graph using KDTree (connect nodes ≤ 40m)
tree = KDTree(coords)
pairs = tree.query_radius(coords, r=40, return_distance=True, sort_results=True)

G = nx.Graph()
for i in range(len(coords)):
    G.add_node(i)
    for j, dist in zip(pairs[0][i][1:], pairs[1][i][1:]):  # Skip self-match
        G.add_edge(i, j, weight=dist)

# Step 3: Cluster via connected components → break large ones using MST + 80m cutoff
cluster_id = 0
df['farm_cluster'] = -1  # Default: noise

for component in nx.connected_components(G):
    if len(component) < 2:
        continue
    subgraph = G.subgraph(component)
    mst = nx.minimum_spanning_tree(subgraph)
    long_edges = [(u, v) for u, v, d in mst.edges(data=True) if d['weight'] > 80]
    mst.remove_edges_from(long_edges)
    for sub_component in nx.connected_components(mst):
        if len(sub_component) >= 2:
            for idx in sub_component:
                df.loc[idx, 'farm_cluster'] = cluster_id
            cluster_id += 1

# Step 4: Summary Output
summary = df[df['farm_cluster'] != -1].groupby('farm_cluster').agg(
    num_houses=('farm_cluster', 'count'),
    mean_latitude=('latitude', 'mean'),
    mean_longitude=('longitude', 'mean'),
    mean_x=('global_x', 'mean'),
    mean_y=('global_y', 'mean')
).reset_index()

# import ace_tools as tools; tools.display_dataframe_to_user(name="Refined Farm Clusters", dataframe=summary)

# Step 5: Plot
plt.figure(figsize=(12, 10))
for label in df['farm_cluster'].unique():
    cluster = df[df['farm_cluster'] == label]
    plt.scatter(cluster['global_x'], cluster['global_y'], s=20, label=f"Cluster {label}" if label != -1 else "Noise")

plt.title("Optimized Graph-Based Poultry Farm Detection")
plt.xlabel("UTM Easting (m)")
plt.ylabel("UTM Northing (m)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.cluster import DBSCAN

def load_house_data(csv_path):
    """
    Load poultry house data in UTM Zone 16N.
    Expects columns: 'latitude' (UTM Northing), 'longitude' (UTM Easting).
    """
    df = pd.read_csv(csv_path)
    if not {'latitude', 'longitude'}.issubset(df.columns):
        raise ValueError("CSV must contain 'latitude' and 'longitude' in UTM coordinates.")
    return df

def cluster_houses(df, eps_house=50, min_samples=2):
    """
    Cluster poultry houses based on distance threshold using DBSCAN.
    """
    coords = df[['longitude', 'latitude']].values
    dbscan = DBSCAN(eps=eps_house, min_samples=min_samples).fit(coords)
    df['initial_cluster'] = dbscan.labels_
    return df

def merge_close_clusters(df, eps_merge=100):
    """
    Merge initial clusters that are spatially close (within eps_merge meters).
    This addresses over-segmentation of farms.
    """
    initial_clusters = df[df['initial_cluster'] != -1].copy()
    
    # Calculate centroids of initial clusters
    centroids = initial_clusters.groupby('initial_cluster')[['longitude', 'latitude']].mean().reset_index()
    
    # Cluster the centroids to merge close clusters
    centroid_coords = centroids[['longitude', 'latitude']].values
    merge_cluster = DBSCAN(eps=eps_merge, min_samples=1).fit(centroid_coords)
    centroids['merged_farm_id'] = merge_cluster.labels_

    # Map merged farm IDs back to each house
    cluster_map = dict(zip(centroids['initial_cluster'], centroids['merged_farm_id']))
    initial_clusters['merged_farm_id'] = initial_clusters['initial_cluster'].map(cluster_map)

    return initial_clusters

def compute_farm_centroids(merged_df):
    """
    Compute centroid and house count for each farm.
    """
    farms = merged_df.groupby('merged_farm_id')[['longitude', 'latitude']].mean().reset_index()
    farms['num_houses'] = merged_df.groupby('merged_farm_id').size().values
    return farms

def save_farms_to_csv(farms_df, output_path):
    """
    Save final farm centroids with house counts to CSV.
    """
    farms_df.to_csv(output_path, index=False)
    print(f"✅ Saved farm centroids to: {output_path}")

# ===================== 🐔 PIPELINE EXECUTION =====================

# Path to your input CSV (replace with your own)
input_csv = "csv/combined_file.csv"  # <-- Input UTM coordinates CSV
output_csv = "csv/refined_poultry_farms.csv"

# Load and process
df = load_house_data(input_csv)
df = cluster_houses(df, eps_house=50, min_samples=2)
merged_df = merge_close_clusters(df, eps_merge=100)
farm_centroids = compute_farm_centroids(merged_df)
save_farms_to_csv(farm_centroids, output_csv)
```
